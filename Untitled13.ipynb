# Question 1: Explain the differences between AI, ML, Deep Learning (DL), and Data Science (DS).


# Answer.
**Artificial Intelligence (AI)**

AI is the broadest concept — it refers to creating systems or machines that can perform tasks that normally require human intelligence.

Goal: Make machines "smart" — able to reason, learn, plan, and solve problems.

Examples: Self-driving cars, chess-playing programs, virtual assistants (Siri, Alexa).

Scope: Includes rule-based systems, machine learning, natural language processing, computer vision, and more.

**Machine Learning (ML)**

Definition: ML is a subset of AI that focuses on learning from data without being explicitly programmed with rules.

Goal: Enable systems to improve performance over time as they get more data.

Techniques: Supervised learning, unsupervised learning, reinforcement learning.

Examples: Spam email detection, movie recommendation systems, credit scoring.

Key Idea: Instead of writing rules manually, the model learns patterns from data.

**Deep Learning (DL)**

Definition: DL is a subset of ML that uses artificial neural networks with multiple layers (deep networks) to model complex patterns.

Goal: Automatically extract features from raw data, especially for unstructured data like images, audio, and text.

Examples: Image recognition (like identifying cats/dogs), voice assistants, autonomous vehicles.

Key Idea: DL works best with large datasets and can automatically learn hierarchical representations.

**Data Science (DS)**

 DS is an interdisciplinary field that focuses on extracting insights and knowledge from data.

Goal: Solve business or scientific problems using data analysis, statistics, and machine learning.

Components:

Data collection & cleaning

Exploratory data analysis (EDA)

Statistical modeling

Machine learning

Data visualization & storytelling

Tools: Python, R, SQL, Excel, Tableau, etc.

Example: Analyzing customer behavior to increase sales or predicting equipment failure in manufacturing.

# Question 2: What are the types of machine learning? Describe each with one real-world example.


# Answer.
**Supervised Learning**

In supervised learning, the model is trained on labeled data — input features with corresponding output labels. The model learns to map inputs to outputs.

Goal: Predict outcomes for new, unseen data.

Techniques: Linear regression, logistic regression, decision trees, random forest, SVM.

Real-world example:

Email spam detection: Input = email content, Output = spam or not spam. The model learns from emails already labeled as spam/non-spam.

**Unsupervised Learning**

 In unsupervised learning, the model is trained on unlabeled data — it tries to find patterns, structures, or groupings in the data without predefined outputs.

Goal: Discover hidden patterns or data structures.

Techniques: K-means clustering, hierarchical clustering, PCA (Principal Component Analysis), anomaly detection.

Real-world example:

Customer segmentation: Grouping customers based on purchasing behavior to identify target segments without pre-labeled categories.

**Semi-Supervised Learning**

 Semi-supervised learning uses a small amount of labeled data and a large amount of unlabeled data. It leverages the labeled data to help classify the unlabeled data.

Goal: Reduce labeling cost while maintaining model accuracy.

Techniques: Self-training, co-training, graph-based methods.

Real-world example:

Medical diagnosis: Only a few medical images are labeled by doctors, and the rest are unlabeled. The model uses both labeled and unlabeled images to improve diagnosis accuracy.

**Reinforcement Learning (RL)**

RL is learning by trial and error, where an agent interacts with an environment and learns to maximize a reward signal over time.

Goal: Learn an optimal strategy or policy to achieve the highest reward.

Techniques: Q-learning, Deep Q Networks (DQN), Policy Gradient methods.

Real-world example:

Self-driving cars: The car learns to drive safely by receiving rewards (staying on road, avoiding collisions) and penalties (accidents or traffic violations).

# Question 3: Define overfitting, underfitting, and the bias-variance tradeoff in machine learning.

#
**verfitting**

Overfitting occurs when a model learns the training data too well, including its noise and random fluctuations, and fails to generalize to new, unseen data.

Signs:

Very high accuracy on training data

Poor accuracy on testing/validation data

Causes:

Model is too complex (too many features, high-degree polynomial, deep neural networks)

Not enough training data

Real-world example:

A model predicting house prices memorizes the exact prices in the training set but performs poorly on new houses because it cannot generalize.

**underfitting**

Underfitting occurs when a model is too simple to capture the underlying patterns in the data. It performs poorly on both training and testing data.

Signs:

Low accuracy on training data

Low accuracy on testing data

Causes:

Model is too simple (linear model for highly nonlinear data)

Not enough features

Real-world example:

Trying to fit a straight line to a highly curved dataset — the model cannot capture the trends.

**Bias-Variance Tradeoff**

The bias-variance tradeoff describes the balance between two sources of error in a model:

Bias: Error due to wrong assumptions in the learning algorithm (underfitting).

Variance: Error due to sensitivity to small fluctuations in the training data (overfitting).

Goal: Minimize total error by balancing bias and variance.

# Question 4: What are outliers in a dataset, and list three common techniques for handling them.

# What are Outliers?

Outliers are data points in a dataset that differ significantly from other observations.

They can be extremely high or low values compared to the rest of the data.

Outliers can skew statistical analyses, affect model performance, and may indicate errors, variability, or rare events.

Real-world examples:

A person with an income of $1,000,000 in a dataset where most incomes are $30,000–$100,000

A student scoring 0 in a test where most scores are 70–90

Sensor readings that are far outside normal operating range

**Common Techniques to Handle Outlier**

a). **Removing Outliers**

Simply remove the outlier rows from the dataset.

Often used when outliers are due to data entry errors.

Example in Python:

import pandas as pd
import numpy as np

df = pd.DataFrame({'score':[50, 55, 60, 200, 65]})
df = df[(np.abs(df['score'] - df['score'].mean()) <= 3*df['score'].std())]

b) **Transforming Data**

Apply log, square root, or Box-Cox transformations to reduce the effect of extreme values.

Helps in normalizing skewed distributions.

Example:

import numpy as np
df['score_log'] = np.log(df['score'])

c) **Imputation / Capping**

Capping (Winsorization): Replace extreme values with the nearest percentile values (e.g., 1st or 99th percentile).

Example in Python:

upper_limit = df['score'].quantile(0.95)
lower_limit = df['score'].quantile(0.05)
df['score'] = np.where(df['score'] > upper_limit, upper_limit, df['score'])
df['score'] = np.where(df['score'] < lower_limit, lower_limit, df['score'])

# Question 5: Explain the process of handling missing values and mention one imputation technique for numerical and one for categorical data.

# Answer.

 Missing values are data points that are not recorded or absent in a dataset.

They can occur due to errors in data collection, entry, or transmission.

Missing values can bias analysis and affect model performance, so handling them properly is crucial.

**Process of Handling Missing Values**

Step 1: Identify Missing Values

Use methods like isnull() or isna() in pandas:

import pandas as pd
df.isnull().sum()

Step 2: Analyze Missing Patterns

Check if missing values are random (MCAR), dependent on other variables (MAR), or systematically missing (MNAR).

This helps decide how to handle them.

Step 3: Handle Missing Values

There are two main approaches:

a) **Removing Missing Values**

Drop rows or columns with missing values if they are few or not critical:

df.dropna(axis=0, inplace=True)  # Drop rows
df.dropna(axis=1, inplace=True)  # Drop columns

b) **Imputation (Filling Missing Values)**

Replace missing values with reasonable estimates based on other data.

```python
#Question 8: Write a Python script to:
# Generate 1000 samples from a normal distribution.
# Introduce 50 random missing values.
# Fill missing values with the column mean.
#Plot a histogram before and after imputation.
#(Include your Python code and output in the code box below.)
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# 1. Generate 1000 samples from a normal distribution
np.random.seed(42)
data = np.random.normal(loc=50, scale=10, size=1000)  # mean=50, std=10

# Convert to DataFrame
df = pd.DataFrame({'value': data})

# 2. Introduce 50 random missing values
missing_indices = np.random.choice(df.index, size=50, replace=False)
df.loc[missing_indices, 'value'] = np.nan

# 3. Plot histogram before imputation
plt.figure(figsize=(12,5))
plt.subplot(1,2,1)
plt.hist(df['value'], bins=30, color='skyblue', edgecolor='black')
plt.title('Histogram Before Imputation')
plt.xlabel('Value')
plt.ylabel('Frequency')

# 4. Fill missing values with the column mean
df['value'].fillna(df['value'].mean(), inplace=True)

plt.subplot(1,2,2)
plt.hist(df['value'], bins=30, color='salmon', edgecolor='black')
plt.title('Histogram After Imputation')
plt.xlabel('Value')
plt.ylabel('Frequency')

plt.tight_layout()
plt.show()
```
