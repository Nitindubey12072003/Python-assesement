# Question 2. How does broadcasting work in NumPy ?

# Answer.
Broadcasting in NumPy is a set of rules that allow arithmetic operations between arrays of different shapes without explicitly copying data.

Instead of looping through elements, NumPy “stretches” one array across the other so their shapes match — without actually creating extra copies in memory.

**How it works :**

When NumPy tries to operate on two arrays:

**1.**Compare shapes from right to left (last dimension first).

**2.**Dimensions are compatible if:

**.**They are equal, or

**.**One of them is 1.

**3.**If compatible, NumPy virtually repeats the smaller array along that dimension to match the larger one.

**4.**If incompatible, it raises an error.

# Question 4. Explain the use of the groupby() method in Pandas .

# Answer.
The groupby() method in Pandas is used to split your data into groups based on some criteria, then apply a function (like sum, mean, count, etc.) to each group, and finally combine the results into a new DataFrame or Series.

It follows the Split → Apply → Combine process.

**Example**

import pandas as pd

data = {
    'Department': ['IT', 'IT', 'HR', 'HR', 'Finance'],

    'Employee': ['John', 'Alice', 'Bob', 'Eve', 'Charlie'],

    'Salary': [50000, 60000, 45000, 48000, 52000]
}


df = pd.DataFrame(data)

 **Group by Department and find average salary**

result = df.groupby('Department')['Salary'].mean()

print(result)

# Question 8. What does the term “vectorized operation” mean in NumPy ?

# Answer.
In NumPy, a vectorized operation means performing a mathematical (or logical) operation on entire arrays at once, without using explicit Python loops to process elements one-by-one.

**Example:**

import numpy as np


a = np.array([1, 2, 3, 4])

b = np.array([10, 20, 30, 40])

result = a + b  # Vectorized addition

print(result)   # [11 22 33 44]

# Question 15. How does NumPy handle multidimensional arrays ?

# Answer.
NumPy handles multidimensional arrays using its ndarray object, which can store data in any number of dimensions (1D, 2D, 3D, and beyond) in a single, continuous memory block for efficiency.

Here’s how it works:

Unified data structure – All arrays, whether 1D (vector), 2D (matrix), or n-dimensional (tensor), are represented by the same ndarray type.

Shape and dimensions – NumPy tracks the number of dimensions (ndim) and the size along each dimension (shape).

Efficient memory layout – Data is stored in a contiguous block, and NumPy uses strides to calculate the position of each element without copying.

Vectorized operations – Mathematical operations apply element-wise to entire arrays, regardless of dimensions, without explicit loops.

Indexing & slicing – Supports powerful ways to access and modify elements, including multi-axis indexing (arr[1, 2]), slicing (arr[:, 0]), and boolean indexing.

Broadcasting – Allows operations between arrays of different shapes by automatically expanding dimensions when possible.

Axes-based operations – Many functions (like sum, mean, max) can operate along specific axes, making it easy to work with rows, columns, or higher-dimensional slices.

# Question 17. A Explain the difference between apply() and map() in Pandas.

# Answer.
In Pandas, both apply() and map() are used to apply functions to data, but they differ in scope and flexibility.

**map()**

 The map() function is designed for element-wise operations on a Pandas Series only; it applies a given function, lambda expression, or mapping dictionary to each individual value in that Series.

 **apply()**

apply() is more versatile—it works on both Series and DataFrames. When used on a Series, it behaves similarly to map(), applying the function element-wise, but when used on a DataFrame, it can apply a function along an axis—either to each column (axis=0) or to each row (axis=1). This makes apply() suitable for more complex operations that require access to multiple values at once.

# Question 21.Why is NumPy’s array slicing faster than Python’s list slicing ?

# Answer.
NumPy's array slicing is faster than Python’s list slicing because NumPy arrays are stored in a single, contiguous block of memory and use views instead of creating full copies when slicing. This means that slicing an array doesn’t duplicate the data—it simply creates a new view object that points to the same underlying memory with adjusted offsets (controlled by strides).

```python
# Q4. How do you calculate the correlation matrix using Seaborn and visualize it with a heatmap?
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Sample DataFrame
data = {
    'Math': [90, 80, 70, 85, 95],
    'Science': [88, 78, 65, 89, 92],
    'English': [75, 85, 80, 70, 90]
}

df = pd.DataFrame(data)

# Calculate correlation matrix
corr_matrix = df.corr()

# Plot heatmap
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt=".2f")

# Add title
plt.title("Correlation Matrix Heatmap")
plt.show()
```
