# 1: What is the difference between descriptive statistics and inferential statistics? Explain with examples.

# Answer.
**1. Descriptive Statistics**

Descriptive statistics are methods used to summarize, organize, and describe the important features of a dataset.

They do not make predictions or generalizations; they only tell you what the data shows.

Techniques used for descriptive statistics :

Measures of central tendency → mean, median, mode .

Measures of dispersion → range, variance, standard deviation

Tables, charts, graphs

**Example:**

Suppose we collect the exam scores of 50 students in a class.

Average score = 72

Highest score = 95

Lowest score = 40

Standard deviation = 10

A histogram shows how the scores are distributed.

 These numbers only describe the class data we have — not beyond.

**2. Inferential Statistics**

Generalizations, or decisions about a larger population.

Purpose: It goes beyond description, using probability and hypothesis testing.

Techniques used:

Estimation (confidence intervals)

Hypothesis testing (t-test, chi-square, ANOVA)

Regression analysis

**Example:**

Instead of collecting exam scores of all students in a university (population), we take a sample of 50 students.

Using their scores, we estimate the average score of all students in the university.

If the sample mean = 72, we might say: “We are 95% confident that the true average score of all students lies between 70 and 74.”

 Here, we are inferring about the entire population from a sample.

# Question 2: What is sampling in statistics? Explain the differences between random and stratified sampling.

# Answer.

Sampling is the process of selecting a subset (sample) from a larger group (population) in order to study the population without examining every member.

**Why we use sampling :**

Studying the entire population is often expensive, time-consuming, or impractical.

A good sample can represent the population and allow us to make valid inferences.

**Example:**

If a company wants to know the average salary of its 10,000 employees, instead of asking everyone, it may randomly select 500 employees and use their data to estimate the population average.

**1. Random Sampling**

 Every member of the population has an equal chance of being selected.

Key idea: No bias in selection.

How it’s done: Lottery method, random number generators, etc.

**Example:**

Suppose you want to survey students in a college with 5,000 students. You randomly pick 200 student IDs using a random number generator.
Each student has the same chance of being included.

**2. Stratified Sampling**

The population is divided into subgroups (strata) based on some characteristics (like age, gender, income, etc.), and then a random sample is taken from each subgroup.

Key idea: Ensures all important groups are represented.

How it’s done: Identify strata → randomly sample from each stratum proportionally.

**Example:**

In the same college survey, suppose 60% of students are undergraduates and 40% are postgraduates.

Divide students into 2 strata: undergraduates & postgraduates.

Randomly select 120 undergraduates and 80 postgraduates (keeping the same 60:40 ratio).
This way, both groups are properly represented.

#Question 3: Define mean, median, and mode. Explain why these measures of central tendency are important.

# Answer.
**1. Mean (Arithmetic Average)**

The mean is the sum of all values divided by the number of values.

Formula:

Mean = Sum of all observations / Number of all observations.


	Example: Test scores = 10, 20, 30, 40, 50

Mean = 10 + 20 + 30 + 40 + 50 / 5
Mean = 30


**2. Median (Middle Value)**

The median is the middle value when data is arranged in ascending or descending order.

If number of observations is odd → middle value

If number of observations is even → average of two middle values

Example: Data = 10, 20, 30, 40, 50
 Median = 30 (middle value)

If Data = 10, 20, 30, 40
 Median = (20+30)/2 = 25

**3. Mode (Most Frequent Value)**

The mode is the value that appears most frequently in the dataset.

A dataset can be unimodal (one mode), bimodal (two modes), or even have no mode.

Example: Data = 10, 20, 20, 30, 40, 50
 Mode = 20 (appears most often).

**Importance of Measures of Central Tendency**

These measures are important because they help us understand the “typical” value in a dataset:

Simplifies large data → Instead of analyzing hundreds of numbers, you can describe the data with just one value (mean, median, or mode).

Comparison → Useful for comparing groups. Example: Comparing average income between two cities.

Decision-making → Businesses, researchers, and policymakers use them to summarize data and guide decisions.

Handles data distribution:

Mean → good for normal distributions.

Median → better when there are outliers/skewed data (e.g., salaries).

Mode → useful for categorical data (e.g., most popular product, favorite color).

# Question 4. Explain skewness and kurtosis. What does a positive skew imply about the data ?

# Answer.
**1. Skewness**

Skewness measures the asymmetry of a data distribution around its mean.

A perfectly symmetrical distribution (like the normal distribution) has skewness = 0.

Types of Skewness:

**Positive Skew (Right-skewed)**

Tail on the right side is longer.

Most values are small, but a few very large values pull the mean to the right.

Order: Mean > Median > Mode

Example: Income distribution (a few very rich people raise the average).

**Negative Skew (Left-skewed)**

Tail on the left side is longer.

Most values are large, but a few very small values pull the mean to the left.

Order: Mean < Median < Mode

Example: Age at retirement (most retire around 60–65, but some retire much earlier).

**2.Kurtosis**

Kurtosis measures the “peakedness” or “tailedness” of a distribution compared to the normal distribution.

It tells us whether data are heavy-tailed (more outliers) or light-tailed (fewer outliers).

Types of Kurtosis:

**1.Mesokurtic (K = 3)**

Normal distribution.

**2. Leptokurtic (K > 3)**

Sharper peak, fatter tails (more extreme values/outliers).

Example: Stock market returns (many extreme values).

**3. Platykurtic (K < 3)**

Flatter peak, thinner tails (fewer outliers).

Example: Uniform distribution.

# Question 5.Implement a Python program to compute the mean, median, and mode of a given list of numbers.
#numbers = [12, 15, 12, 18, 19, 12, 20, 22, 19, 19, 24, 24, 24, 26, 28]

```python
import statistics
numbers = [12, 15, 12, 18, 19, 12, 20, 22, 19, 19, 24, 24, 24, 26, 28]
mean_value = statistics.mean(numbers)
median = statistics.median(numbers)
mode = statistics.mode(numbers)
print("Mean:",mean_value)
print("Median:",median)
print("Mode:",mode)
```

# Question 5. Compute the covariance and correlation coefficient between the following  two datasets provided as lists in Python:
 # list_x = [10, 20, 30, 40, 50]
# list_y = [15, 25, 35, 45, 60]

```python
import numpy as np
list_x = [10, 20, 30, 40, 50]
list_y = [15, 25, 35, 45, 60]

# Convert to numpy arrays
x = np.array(list_x)
y = np.array(list_y)

# covariance matrix
cov_matrix = np.cov(x, y, bias=False)
covariance = cov_matrix[0,1]

# Correlation coefficient
correlation_matrix = np.corrcoef(x,y)
correlation = correlation_matrix[0,1]
print("Covariance :",covariance)
print("Correlation coefficient :",correlation)

```

# Write a Python script to draw a boxplot for the following numeric list and identify its outliers. Explain the result:
#data = [12, 14, 14, 15, 18, 19, 19, 21, 22, 22, 23, 23, 24, 26, 29, 35]

```python
import matplotlib.pyplot as plt
import seaborn as sns
# Given data
data = [12, 14, 14, 15, 18, 19, 19, 21, 22, 22, 23, 23, 24, 26, 29, 35]
# Draw plot
plt.figure(figsize=(6,4))
sns.boxplot(data)
plt.title("Boxplot of Data with Outlier Detection")
plt.show()
# --- Outlier detection manually ---
import numpy as np

Q1 = np.percentile(data, 25)  # First quartile
Q3 = np.percentile(data, 75)  # Third quartile
IQR = Q3 - Q1
# Define bounds for outliers
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
outliers = [x for x in data if x < lower_bound or x > upper_bound]
print("Q1 =", Q1)
print("Q3 =", Q3)
print("IQR =", IQR)
print("Lower Bound =", lower_bound)
print("Upper Bound =", upper_bound)
print("Outliers:", outliers)
```

# Question 8: You are working as a data analyst in an e-commerce company.
#The marketing team wants to know if there is a relationship between
#advertising spend anddaily sales.● Explain how you would use covariance and

#correlation to explore thisrelationship.Write Python code to compute the

#correlation between the two lists: advertising_spend = [200, 250, 300, 400, 500]

#daily_sales = [2200, 2450, 2750, 3200, 4000]

```python
import numpy as np

# Given data
advertising_spend = [200, 250, 300, 400, 500]
daily_sales = [2200, 2450, 2750, 3200, 4000]

# Convert to numpy arrays
x = np.array(advertising_spend)
y = np.array(daily_sales)

# Covariance matrix
cov_matrix = np.cov(x, y, bias=False)
covariance = cov_matrix[0, 1]

# Correlation coefficient
correlation_matrix = np.corrcoef(x, y)
correlation = correlation_matrix[0, 1]

print("Covariance:", covariance)
print("Correlation Coefficient:", correlation)
```
