# Q1. What is a RESTful API ?

# Answer.
A RESTful API (Representational State Transfer API) is an interface between clients and servers that follows the principles of the REST architecture style. It lets applications communicate with each other over HTTP by treating data as resources that can be created, read, updated, or deleted (CRUD).

# Question 8. What is the difference between GET and POST HTTP methods ?

# Answer.
**1.** **GET**

**.** Retrieve or read data from the server.

**.** Data is appended to the URL as query parameters (?key=value).

**.** Parameters are visible in the URL → Not secure for sensitive data.

**2.** **POST**

**.** Send data to the server to create a new resource.

**.** Data is sent in the request body (JSON, form data, etc.).

**.** Parameters are hidden in the request body → Safer for sensitive data.

# Question 14. How do you create a RESTful API endpoint using Flask ?

# Answer.

from flask import Flask, request, jsonify

app = Flask(__name__)

users = [{"id": 1, "name": "Nitin"}]


GET: Retrieve all users

@app.route("/users", methods=["GET"])

def get_users():

    return jsonify(users)

GET: Retrieve a specific user

@app.route("/users/<int:user_id>", methods=["GET"])

def get_user(user_id):

    user = next((u for u in users if u["id"] == user_id), None)

    if not user:

        return jsonify({"error": "User not found"}), 404
    return jsonify(user)

POST: Create a new user

@app.route("/users", methods=["POST"])

def add_user():

    new_user = request.get_json()

    users.append(new_user)

    return jsonify(new_user), 201

PUT: Update a user

@app.route("/users/<int:user_id>", methods=["PUT"])

def update_user(user_id):

    data = request.get_json()

    user = next((u for u in users if u["id"] == user_id), None)

    if not user:

        return jsonify({"error": "User not found"}), 404

    user.update(data)

    return jsonify(user)

DELETE: Remove a user

@app.route("/users/<int:user_id>", methods=["DELETE"])

def delete_user(user_id):

    global users

    users = [u for u in users if u["id"] != user_id]

    return jsonify({"message": "User deleted"})

Step 4: Run the Application

if __name__ == "__main__":

    app.run(debug=True)

```python
# 6. How do you handle forms in Flask ?
from flask import Flask, render_template, request

app = Flask(__name__)

@app.route('/')
def index():
    return render_template('form.html')
@app.route('/submit', methods=['POST'])
def submit():
    name = request.form['name']  # Access form data
    age = request.form['age']
    return f"Hello {name}, you are {age} years old!"
@app.route('/search', methods=['GET'])
def search():
    query = request.args.get('q')  # Access query parameters
    return f"You searched for: {query}"
if __name__ == '__main__':
    app.run(debug=True)
```

```python
# 7.  How can you validate form data in Flask ?
from flask import Flask, request, render_template

app = Flask(__name__)

@app.route('/')
def form():
    return render_template('form.html')  # simple HTML form

@app.route('/submit', methods=['POST'])
def submit():
    name = request.form.get('name')
    age = request.form.get('age')

    # Basic validation
    if not name or not age:
        return "Error: Name and age are required!", 400

    try:
        age = int(age)
        if age <= 0:
            return "Error: Age must be positive!", 400
    except ValueError:
        return "Error: Age must be a number!", 400

    return f"Hello {name}, you are {age} years old!"
if __name__ == '__main__':
    app.run(debug=True)
```

```python
# Q8. How do you manage sessions in Flask ?
from flask import Flask, session

app = Flask(__name__)
app.secret_key = "supersecret123"  # keep this safe in production
@app.route('/')
def index():
    if 'username' in session:
        return f"Logged in as {session['username']}"
    return "You are not logged in"

@app.route('/login', methods=['POST', 'GET'])
def login():
    if request.method == 'POST':
        session['username'] = request.form['username']  # store in session
        return redirect(url_for('index'))
    return '''
        <form method="post">
            <input type="text" name="username">
            <input type="submit" value="Login">
        </form>
    '''

@app.route('/logout')
def logout():
    session.pop('username', None)  # remove from session
    return redirect(url_for('index'))
if __name__ == '__main__':
    app.run(debug=True)
```

```python
# Q10. How do you handle errors in Flask (e.g., 404) ?
from flask import Flask, render_template

app = Flask(__name__)

@app.errorhandler(404)
def page_not_found(error):
    return render_template('404.html'), 404

@app.errorhandler(500)
def internal_error(error):
    return render_template('500.html'), 500
if __name__ == '__main__':
    app.run(debug=True)
```

```python
#Q11. How do you structure a Flask app using Blueprints ?
from flask import Blueprint, render_template, request, redirect, url_for

auth_bp = Blueprint('auth', __name__, template_folder='templates')

@auth_bp.route('/login', methods=['GET', 'POST'])
def login():
    if request.method == 'POST':
        username = request.form['username']
        return redirect(url_for('auth.dashboard', user=username))
    return render_template('auth/login.html')

@auth_bp.route('/dashboard/<user>')
def dashboard(user):
    return f"Welcome, {user}!"
if __name__ == '__main__':
    app.run(debug=True)

```

```python
#Q12.How do you define a custom Jinja filter in Flask?
from flask import Flask

app = Flask(__name__)

# Define a custom filter function
def reverse_string(s):
    return s[::-1]

# Register the filter with Jinja
app.jinja_env.filters['reverse'] = reverse_string

if __name__ == '__main__':
    app.run(debug=True)
```

```python
# Q13. How can you redirect with query parameters in Flask?
from flask import Flask, redirect, url_for, request

app = Flask(__name__)

@app.route('/')
def home():
    return "Home Page"

@app.route('/search')
def search():
    query = request.args.get('q')
    return f"You searched for: {query}"

@app.route('/go')
def go():
    # Redirect to /search?q=flask
    return redirect(url_for('search', q='flask'))
if __name__ == '__main__':
    app.run(debug=True)
```
