# Question 1: What is hypothesis testing in statistics?


# Answer
Hypothesis testing is a statistical method used to make decisions or inferences about a population based on sample data. It helps to determine whether there is enough evidence to support or reject a specific assumption (hypothesis) about the population.

**Key Components:**

**1. Null Hypothesis (H₀):**

It represents the default assumption or no effect. For example, “There is no difference between two groups.”

**2. Alternative Hypothesis (H₁ or Ha):**

It is what you want to prove, such as “There is a significant difference between the two groups.”

**3. Test Statistic:**

A calculated value from the sample data used to determine how extreme the observed data is.

**4. P-value:**

The probability of observing the data if the null hypothesis is true. A small p-value (usually less than 0.05) indicates strong evidence against the null hypothesis.

**5. Significance Level (α):**

A threshold (commonly 0.05) below which you reject the null hypothesis.

**6. Decision:**

Reject H₀ if the p-value ≤ α → evidence supports the alternative hypothesis.

Fail to reject H₀ if the p-value > α → insufficient evidence to support the alternative hypothesis.

# Question 2: What is the null hypothesis, and how does it differ from the alternative hypothesis?


# Answer.
**Null Hypothesis (H₀):**

The null hypothesis is a statement that assumes no effect, no difference, or no relationship between variables.

It represents the default or status quo position that there is nothing new happening.

It is what you try to disprove or test against.

**Example:** “There is no difference in average test scores between two teaching methods.”

**Alternative Hypothesis (H₁ or Ha):**

The alternative hypothesis is a statement that assumes there is an effect, difference, or relationship.

It represents what you are trying to prove or support with data.

**Example:** “There is a difference in average test scores between two teaching methods.”

# Question 3: Explain the significance level in hypothesis testing and its role in deciding the outcome of a test.


# Answer.
The significance level, denoted by α (alpha), is a threshold that defines how much risk of error you're willing to accept when making a decision in hypothesis testing.

Specifically, it is:

The probability of rejecting the null hypothesis (H₀) when it is actually true.

Also known as the Type I error rate.

Common values for α are:

0.05 (5%) – most widely used.

0.01 (1%) – for more strict testing.

0.10 (10%) – for less strict scenarios.

**Role in Deciding the Outcome of a Test:**

Compare the p-value with α:

If p-value ≤ α → reject the null hypothesis (H₀). The result is statistically significant.

If p-value > α → fail to reject the null hypothesis. There isn’t enough evidence to support the alternative hypothesis.

Controls how cautious or bold you are:

A smaller α (e.g., 0.01) means you're being more conservative and require stronger evidence to reject H₀.

A larger α (e.g., 0.10) means you're more willing to take a risk in rejecting H₀.

Helps manage errors:

Type I error (false positive): Rejecting H₀ when it’s true → controlled by α.

Type II error (false negative): Failing to reject H₀ when it’s false → affected indirectly by sample size, effect size, and α.

# Question 4: What are Type I and Type II errors? Give examples of each.


# Answer.
In hypothesis testing, errors can occur because decisions are made based on sample data rather than the entire population. These errors are classified into two types:

**Type I Error (False Positive):**

Occurs when the null hypothesis (H₀) is true, but we incorrectly reject it.

It means we conclude that an effect or difference exists when it actually doesn't.

The probability of making a Type I error is denoted by α (alpha), the significance level.

Example:
A medical test concludes that a patient has a disease when, in fact, the patient is healthy.

**Type II Error (False Negative):**

Occurs when the null hypothesis (H₀) is false, but we fail to reject it.

It means we conclude there is no effect or difference when one actually exists.

The probability of making a Type II error is denoted by β (beta).

Example:
A medical test fails to detect a disease when the patient actually has it.

# Question 5: What is the difference between a Z-test and a T-test ? Explain when to use each.

# Answer.
A Z-test and a T-test are both statistical methods used to compare sample means and determine if observed differences are significant, but they differ in their assumptions and usage.

**Z-test** is applied when the sample size is large (typically greater than 30) and the population variance or standard deviation is known. It uses the standard normal distribution to calculate probabilities and is suitable when the data follows a normal distribution or when the sample size is large enough for the Central Limit Theorem to apply. On the other hand, a T-test is used when the sample size is small (usually 30 or less) and the population variance is unknown, requiring estimation from the sample itself. The T-test relies on the Student’s t-distribution, which adjusts for smaller sample sizes by incorporating degrees of freedom.

**T-test** is appropriate when working with small samples or when the population parameters are not available. In summary, a Z-test is ideal for large samples with known variance, while a T-test is best suited for smaller samples where variance must be inferred from the data. Choosing the correct test ensures that conclusions drawn from the data are statistically sound and reliable.

# Question 6: Write a Python program to generate a binomial distribution with n=10 and p=0.5, then plot its histogram.


```python
import numpy as np
import matplotlib.pyplot as plt

# Parameters for the binomial distribution
n = 10      # number of trials
p = 0.5     # probability of success
size = 1000  # number of samples

# Generate the binomial distribution
data = np.random.binomial(n, p, size)

# Plot the histogram
plt.hist(data, bins=range(n+2), edgecolor='black', align='left')
plt.title('Binomial Distribution (n=10, p=0.5)')
plt.xlabel('Number of Successes')
plt.ylabel('Frequency')
plt.xticks(range(n+1))
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

# Question 7: Implement hypothesis testing using Z-statistics for a sample dataset in Python. Show the Python code and interpret the results. sample_data = [49.1, 50.2, 51.0, 48.7, 50.5, 49.8, 50.3, 50.7, 50.2, 49.6,50.1, 49.9, 50.8, 50.4, 48.9, 50.6, 50.0, 49.7, 50.2, 49.5,50.1, 50.3, 50.4, 50.5, 50.0, 50.7, 49.3, 49.8, 50.2, 50.9,50.3, 50.4, 50.0, 49.7, 50.5, 49.9]


```python
import numpy as np
from scipy.stats import norm

# Sample data
sample_data = [
    49.1, 50.2, 51.0, 48.7, 50.5, 49.8, 50.3, 50.7, 50.2, 49.6,
    50.1, 49.9, 50.8, 50.4, 48.9, 50.6, 50.0, 49.7, 50.2, 49.5,
    50.1, 50.3, 50.4, 50.5, 50.0, 50.7, 49.3, 49.8, 50.2, 50.9,
    50.3, 50.4, 50.0, 49.7, 50.5, 49.9
]

# Known population parameters
mu = 50.0        # population mean
sigma = 0.5      # population standard deviation

# Sample statistics
sample_mean = np.mean(sample_data)
n = len(sample_data)

# Calculate the Z statistic
z_stat = (sample_mean - mu) / (sigma / np.sqrt(n))

# Calculate the two-tailed p-value
p_value = 2 * (1 - norm.cdf(abs(z_stat)))

# Print results
print(f"Sample Mean = {sample_mean:.4f}")
print(f"Z-Statistic = {z_stat:.4f}")
print(f"P-value = {p_value:.4f}")

# Interpretation
alpha = 0.05
if p_value < alpha:
    print("Reject the null hypothesis: The sample mean is significantly different from the population mean.")
else:
    print("Fail to reject the null hypothesis: There is no significant difference from the population mean.")
```

# Question 8: Write a Python script to simulate data from a normal distribution and calculate the 95% confidence interval for its mean. Plot the data using Matplotlib.


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

# Simulate data from a normal distribution
np.random.seed(0)  # for reproducibility
mu = 50            # population mean
sigma = 5          # population standard deviation
n = 100            # sample size

# Generate the sample data
data = np.random.normal(mu, sigma, n)

# Calculate sample mean and standard error
sample_mean = np.mean(data)
sample_std = np.std(data, ddof=1)  # sample standard deviation
std_error = sample_std / np.sqrt(n)

# Calculate the 95% confidence interval
confidence_level = 0.95
degrees_freedom = n - 1
confidence_interval = stats.t.interval(confidence_level, degrees_freedom, loc=sample_mean, scale=std_error)

# Print the results
print(f"Sample Mean = {sample_mean:.4f}")
print(f"95% Confidence Interval = ({confidence_interval[0]:.4f}, {confidence_interval[1]:.4f})")

# Plot the histogram
plt.hist(data, bins=15, edgecolor='black', alpha=0.7)
plt.axvline(sample_mean, color='red', linestyle='--', label=f'Mean: {sample_mean:.2f}')
plt.axvline(confidence_interval[0], color='green', linestyle='--')
plt.axvline(confidence_interval[1], color='green', linestyle='--', label='95% Confidence Interval')
plt.title('Simulated Normal Distribution with 95% Confidence Interval')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.legend()
plt.show()
```

# Question 9: Write a Python function to calculate the Z-scores from a dataset and visualize the standardized data using a histogram. Explain what the Z-scores represent in terms of standard deviations from the mean.

```python
import numpy as np
import matplotlib.pyplot as plt

def calculate_and_plot_z_scores(data):
    # Calculate mean and standard deviation
    mean = np.mean(data)
    std = np.std(data, ddof=1)  # sample standard deviation

    # Calculate Z-scores
    z_scores = (data - mean) / std

    # Plot histogram of Z-scores
    plt.hist(z_scores, bins=15, edgecolor='black', alpha=0.7)
    plt.title('Histogram of Z-scores')
    plt.xlabel('Z-score')
    plt.ylabel('Frequency')
    plt.axvline(0, color='red', linestyle='--', label='Mean (Z=0)')
    plt.legend()
    plt.grid(axis='y', linestyle='--', alpha=0.7)
    plt.show()

    return z_scores

# Example usage
np.random.seed(42)  # for reproducibility
sample_data = np.random.normal(50, 5, 100)  # dataset with mean=50, std=5, n=100

z_scores = calculate_and_plot_z_scores(sample_data)

# Print first 5 Z-scores
print("First 5 Z-scores:", z_scores[:5])
```
