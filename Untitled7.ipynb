#Question 1. What is the difference between interpreted and compiled languages?


#Answer.
**1. Compiled Languages**

Process: The entire source code is translated at once into machine code by a compiler before execution.

Output: Produces a separate executable file (e.g., .exe in Windows).

Execution Speed: Faster at runtime because the code is already compiled.

Examples: C, C++, Rust, Go

**2. Interpreted Languages**

Process: The source code is translated line-by-line at runtime by an interpreter.

Output: No separate executable is generated; it runs directly.

Execution Speed: Slower compared to compiled languages (due to real-time interpretation).

Examples: Python, JavaScript, Ruby, PHP

#Question 9. What is the difference between multithreading and multiprocessing ?

#Answer.
The main difference between multithreading and multiprocessing lies in how they achieve concurrency and how they use system resources like CPU cores and memory.

**Multithreading**

 Running multiple threads (smaller units of a process) within a single process.

Resource Use: All threads share the same memory space.

Best For: I/O-bound tasks (e.g., file read/write, network requests).

Limitation in Python: Due to the Global Interpreter Lock (GIL), only one thread runs at a time in CPython (Python’s main implementation), limiting true parallelism for CPU-bound tasks.

Faster Context Switching: Because threads share memory, switching between them is quick.

**Multiprocessing**

 Running multiple separate processes, each with its own Python interpreter and separate memory space.

Resource Use: Each process gets its own memory and runs independently.

Best For: CPU-bound tasks (e.g., complex computations, data processing).

True Parallelism: Takes full advantage of multiple CPU cores.

Heavier Than Threads: More memory consumption and slower context switching compared to threads.

#Question 18.What is the difference between os.fork() and multiprocessing in Python ?

#Answer.
Both os.fork() and the multiprocessing module are used to create new processes, but they work differently and are suited for different use cases.

**1.** os.fork()

Low-level function (available on Unix/Linux only).

Creates a child process by duplicating the current process.

Both parent and child continue from the same point in the code, but with different process IDs.

You must manually manage communication, synchronization, and process handling.

**2.** Multiprocessing Module

High-level module built on top of os.fork() (on Unix) or spawn() (on Windows).

Provides a clean, cross-platform way to create and manage processes.

Handles inter-process communication, synchronization, and shared data easily.

```python
# 6.Write a Python program that logs an error message to a log file when a division by zero exception occurs.
import logging

# Configure logging to write to a file
logging.basicConfig(filename="error.log", level=logging.ERROR,
                    format='%(asctime)s - %(levelname)s - %(message)s')

try:
    a = int(input("Enter numerator: "))
    b = int(input("Enter denominator: "))
    result = a / b
    print("Result:", result)
except ZeroDivisionError as e:
    print("Error: Division by zero is not allowed.")
    logging.error("Division by zero attempted: %s", e)
```

```python
# 18. How would you implement a basic logging setup that logs to a file with rotation after 1MB .
import logging
from logging.handlers import RotatingFileHandler

# Set up a rotating log handler
log_handler = RotatingFileHandler(
    filename="app.log",       # Log file name
    maxBytes=1_000_000,       # Rotate after 1MB
    backupCount=3             # Keep last 3 log files
)

# Format for log messages
formatter = logging.Formatter('%(asctime)s - %(levelname)s - %(message)s')
log_handler.setFormatter(formatter)

# Set up the logger
logger = logging.getLogger("MyLogger")
logger.setLevel(logging.INFO)        # Set the minimum level to log
logger.addHandler(log_handler)

# Example usage
for i in range(10000):
    logger.info(f"This is log message #{i}")
```

```python
# 21. Write a Python program that reads a file and prints the number of occurrences of a specific word
# Ask user for filename and word to search
filename = input("Enter the filename: ")
search_word = input("Enter the word to count: ")

try:
    with open(filename, "r") as file:
        content = file.read().lower()  # Convert to lowercase for case-insensitive search
        word_list = content.split()    # Split text into words
        count = word_list.count(search_word.lower())
        print(f"The word '{search_word}' appears {count} times in '{filename}'.")
except FileNotFoundError:
    print(f"Error: The file '{filename}' was not found.")
except Exception as e:
    print(f"An unexpected error occurred: {e}")
```

```python
# 22. How can you check if a file is empty before attempting to read its contents ?
import os

filename = "example.txt"

if os.path.exists(filename):
    if os.path.getsize(filename) == 0:
        print("The file is empty.")
    else:
        with open(filename, "r") as file:
            content = file.read()
            print("File content:\n", content)
else:
    print(f"File '{filename}' does not exist.")
```

```python
# 23. Write a Python program that writes to a log file when an error occurs during file handling.
import logging

# Configure logging to write error messages to a log file
logging.basicConfig(
    filename='file_errors.log',
    level=logging.ERROR,
    format='%(asctime)s - %(levelname)s - %(message)s'
)

# Ask user for the file to open
filename = input("Enter the filename to read: ")

try:
    with open(filename, "r") as file:
        content = file.read()
        print("File content:\n", content)

except FileNotFoundError as e:
    print(f"Error: The file '{filename}' was not found.")
    logging.error("FileNotFoundError: %s", e)

except PermissionError as e:
    print(f"Error: Permission denied for file '{filename}'.")
    logging.error("PermissionError: %s", e)

except Exception as e:
    print("An unexpected error occurred.")
    logging.error("Unexpected error: %s", e)
```
