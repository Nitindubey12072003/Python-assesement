# Q1. Explain the fundamental differences between DDL, DML, and DQL commands in SQL. Provide one example for each type of command.

# Answer.
**1.** DDL – Data Definition Language Purpose:

DDL commands are used to define, create, or modify the structure of database objects such as tables, views, indexes, etc.

Key Characteristics:

Works on schema (structure), not data.

Auto-commits (changes cannot be rolled back in most systems).

Affects the entire table or database object.

Examples of DDL Commands:

CREATE, ALTER, DROP, TRUNCATE

Example:
CREATE TABLE Students (
    student_id INT,
    name VARCHAR(50),
    age INT
);

**2.** DML – Data Manipulation Language
Purpose:

DML commands are used to insert, update, or delete data stored in database tables.

Key Characteristics:

Works on data inside tables.

Changes can be rolled back (transactional).

Does not affect table structure.

Examples of DML Commands:

INSERT, UPDATE, DELETE

Example:
INSERT INTO Students (student_id, name, age)
VALUES (1, 'John Doe', 20);

 **3.** DQL – Data Query Language
Purpose:

DQL commands are used to retrieve data from database tables.

Key Characteristics:

Only reads data; does not modify it.

Produces result sets based on conditions.

Example of DQL Command:

SELECT

Example:
SELECT name, age FROM Students WHERE age > 18;

#Q2. What is the purpose of SQL constraints? Name and describe three common types of constraints, providing a simple scenario where each would be useful.


# Answer.
Three Common Types of SQL Constraints (with scenarios) :

**1.** PRIMARY KEY Constraint
Purpose:

Ensures that each row in a table can be uniquely identified.
A primary key:

Must be unique

Cannot be NULL

Scenario:

In a Students table, student_id must uniquely identify each student.

Example:
CREATE TABLE Students (
    student_id INT PRIMARY KEY,
    name VARCHAR(50)
);

**2.** FOREIGN KEY Constraint
Purpose:

Enforces a relationship between two tables.
It ensures that a value in one table must exist in another table.

Scenario:

An Orders table should only contain customer_id values that exist in the Customers table.

Example:
CREATE TABLE Orders (
    order_id INT PRIMARY KEY,
    customer_id INT,
    FOREIGN KEY (customer_id) REFERENCES Customers(customer_id)
);

**3.** UNIQUE Constraint
Purpose:

Ensures that the values in a column are not duplicated, but unlike a primary key, a UNIQUE column can contain NULL (depending on the SQL database).

Scenario:

In a UserAccounts table, the email address must be unique for each user.

Example:
CREATE TABLE UserAccounts (
    user_id INT PRIMARY KEY,
    email VARCHAR(100) UNIQUE
);

# Q3. Explain the difference between LIMIT and OFFSET clauses in SQL. Howwould you use them together to retrieve the third page of results, assuming each pagehas 10 records ?

# Answer.
 **Difference Between LIMIT and OFFSET in SQL LIMIT :**

Specifies how many rows to return.

Controls the number of records in the output.

**OFFSET**

Specifies how many rows to skip before starting to return records.

Controls the starting point of the output.

using LIMIT and OFFSET Together

These two clauses are often used together for pagination (showing records page by page).

 Example: Retrieve the third page of results

Each page has 10 records.

✔ Page 1 → records 1–10
✔ Page 2 → records 11–20
✔ Page 3 → records 21–30

To retrieve page 3:

OFFSET = (page_number − 1) × page_size
→ (3 − 1) × 10 = 20

LIMIT = page_size = 10

 SQL Query:

SELECT *
FROM table_name
LIMIT 10 OFFSET 20;

# Q4. What is a Common Table Expression (CTE) in SQL, and what are its main benefits? Provide a simple SQL example demonstrating its usage.


# Answer.
A CTE:

Is like a temporary table created within a query.

Can be referenced multiple times within the main query.

Improves readability, maintainability, and sometimes performance.

 Main Benefits of Using CTEs
1. Improved Readability

CTEs allow you to structure queries in a cleaner and more understandable way compared to nested subqueries.

2. Reusability

You can reference the CTE multiple times in the main query without rewriting the same logic.

3. Recursive Queries

CTEs support recursion, which is useful for hierarchical data (e.g., employee → manager tables, folder structures).

4. Easier Debugging

You can isolate parts of large queries as separate CTE blocks.

# Q5. Describe the concept of SQL Normalization and its primary goals. Briefly explain the first three normal forms (1NF, 2NF, 3NF).


# Answer.
SQL Normalization is a process used in database design to organize data efficiently and eliminate problems such as redundancy, inconsistency, and dependency anomalies.

 Primary Goals of Normalization

Reduce data redundancy
— Avoid storing the same data in multiple places.

Ensure data integrity
— Prevent inconsistent or duplicate information.

Improve data organization
— Structure data in logical, related tables.

Simplify maintenance
— Updates or deletions affect only necessary data.

Avoid anomalies
— Prevent update, insert, and delete anomalies.

```python
CREATE DATABASE ECommerceDB;
USE ECommerceDB;
CREATE TABLE Categories (
    CategoryID INT PRIMARY KEY,
    CategoryName VARCHAR(50) NOT NULL UNIQUE
);
CREATE TABLE Products (
    ProductID INT PRIMARY KEY,
    ProductName VARCHAR(100) NOT NULL UNIQUE,
    CategoryID INT,
    Price DECIMAL(10,2) NOT NULL,
    StockQuantity INT,
    FOREIGN KEY (CategoryID) REFERENCES Categories(CategoryID)
);
CREATE TABLE Customers (
    CustomerID INT PRIMARY KEY,
    CustomerName VARCHAR(100) NOT NULL,
    Email VARCHAR(100) UNIQUE,
    JoinDate DATE
);
CREATE TABLE Orders (
    OrderID INT PRIMARY KEY,
    CustomerID INT,
    OrderDate DATE NOT NULL,
    TotalAmount DECIMAL(10,2),
    FOREIGN KEY (CustomerID) REFERENCES Customers(CustomerID)
);
INSERT INTO Categories (CategoryID, CategoryName)
VALUES
(1, 'Electronics'),
(2, 'Books'),
(3, 'Home Goods'),
(4, 'Apparel');
```

```python
SELECT
    c.CustomerName,
    c.Email,
    COUNT(o.OrderID) AS TotalNumberOfOrders
FROM Customers c
LEFT JOIN Orders o
    ON c.CustomerID = o.CustomerID
GROUP BY
    c.CustomerID, c.CustomerName, c.Email
ORDER BY
    c.CustomerName;
```

```python
SELECT
    p.ProductName,
    p.Price,
    p.StockQuantity,
    c.CategoryName
FROM Products p
INNER JOIN Categories c
    ON p.CategoryID = c.CategoryID
ORDER BY
    c.CategoryName ASC,
    p.ProductName ASC;
```

```python
WITH RankedProducts AS (
    SELECT
        c.CategoryName,
        p.ProductName,
        p.Price,
        ROW_NUMBER() OVER (
            PARTITION BY c.CategoryName
            ORDER BY p.Price DESC
        ) AS rn
    FROM Products p
    JOIN Categories c
        ON p.CategoryID = c.CategoryID
)
SELECT
    CategoryName,
    ProductName,
    Price
FROM RankedProducts
WHERE rn <= 2
ORDER BY CategoryName, Price DESC;
```
