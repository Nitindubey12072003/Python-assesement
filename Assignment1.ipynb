```python
# Write a function to demonstrate the behavior of mutable and immutable arguments.
# Function to modify arguments
def modify(a, b):
    print("\nInside function before modification:")
    print("a =", a)
    print("b =", b)

    # Trying to modify both arguments
    a = a + 10          # This will not affect the original variable (immutable)
    b.append(100)       # This will affect the original list (mutable)

    print("\nInside function after modification:")
    print("a =", a)
    print("b =", b)

# Main code
x = 5          # Immutable (int)
y = [1, 2, 3]  # Mutable (list)

print("Before calling the function:")
print("x =", x)
print("y =", y)

modify(x, y)

print("\nAfter calling the function:")
print("x =", x)  # No change, because int is immutable
print("y =", y)  # Changed, because list is mutable
```

```python
#Write a program that performs basic arithmetic operations on two user-input numbers.
# Take two numbers as input from the user
num1 = float(input("Enter the first number: "))
num2 = float(input("Enter the second number: "))

# Perform arithmetic operations
addition = num1 + num2
subtraction = num1 - num2
multiplication = num1 * num2

# Handle division by zero
if num2 != 0:
    division = num1 / num2
else:
    division = "Error! Division by zero."

# Display the results
print("\nResults:")
print("Addition:", addition)
print("Subtraction:", subtraction)
print("Multiplication:", multiplication)
print("Division:", division)
```

```python
# Write a program to demonstrate the use of logical operators.
# Take input from the user
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))

# Using 'and' operator
if a > 0 and b > 0:
    print("Both numbers are positive.")
else:
    print("At least one number is not positive.")

# Using 'or' operator
if a == 0 or b == 0:
    print("At least one number is zero.")
else:
    print("Neither number is zero.")

# Using 'not' operator
is_negative = a < 0
if not is_negative:
    print("The first number is not negative.")
else:
    print("The first number is negative.")
```

```python
# Write a Python program to convert user input from string to integer, float, and boolean types.


# Take input from the user
user_input = input("Enter a value: ")

# Convert to integer
try:
    int_value = int(user_input)
    print("Integer value:", int_value)
except ValueError:
    print("Cannot convert to integer.")

# Convert to float
try:
    float_value = float(user_input)
    print("Float value:", float_value)
except ValueError:
    print("Cannot convert to float.")

# Convert to boolean
bool_value = bool(user_input)  # Any non-empty string is True
print("Boolean value:", bool_value)
```

```python
#Write a Python program to calculate the factorial of a number provided by the user using a while loop.
# Take input from the user
num = int(input("Enter a number to find its factorial: "))

# Initialize variables
factorial = 1
i = 1

# Check if the number is negative
if num < 0:
    print("Factorial is not defined for negative numbers.")
else:
    while i <= num:
        factorial *= i
        i += 1
    print(f"The factorial of {num} is: {factorial}")
```
