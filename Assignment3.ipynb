Q1.What is the difference between a function and a method in Python ?

Answer.

In Python, functions and methods are used to perform actions, but they differ in how they are defined and called.

#1. Function

.A function is a block of reusable code that performs a specific task.

.It is defined using the def keyword.

.Functions are not associated with any object.

.You call a function directly using its name.

**Example**

def greet(name):

    return f"Hello, {name}!"

print(greet("Alice"))


#2. Method

.A method is a function that is associated with an object (usually an instance of a class).

.It is called on an object, and the object itself is implicitly passed as the first argument (self).

Defined inside a class.

**Example**:

class Greeter:

    def greet(self, name):

        return f"Hello, {name}!"

g = Greeter()

print(g.greet("Alice"))   #Calling a method

Q5.What are iterators in Python and how do they differ from iterables?

Answer.

In Python, iterators and iterables are related concepts used for looping and accessing elements one at a time, but they are not the same:

#Iterables

An iterable is any Python object capable of returning its members one at a time.

Examples: list, tuple, set, str, dict, and any object that implements the __iter__() method.

You can loop through an iterable using a for loop.

**Example of an iterable**

my_list = [1, 2, 3]

for item in my_list:

print(item)




#Iterators

An iterator is an object with a __next__() method that returns the next item in the sequence.

Once all items are consumed, it raises a StopIteration exception.

Iterators remember their current position.

**Example**

my_iter = iter([1, 2, 3])

print(next(my_iter))  # 1

print(next(my_iter))  # 2

print(next(my_iter))  # 3

#next(my_iter)  #Raises Stop Iteration




Q6. Explain the concept of generators in Python and how they are defined.

Answer.

A generator in Python is a special type of iterator that is used to generate values on the fly (i.e., one at a time and only when needed), using less memory than regular lists.

Key Concepts:

Generators yield values one at a time.

They are lazy iterators: they do not compute all items at once.

They maintain state between successive calls automatically.

**Example:**

How to Define a Generator:

You define a generator using a function with the yield keyword instead of return.

**Syntax**:

def my_generator():

    yield 1
    yield 2
    yield 3
This function does not return values immediately. Instead, it returns a generator object, which can be iterated.

Using the Generator:

gen = my_generator()

print(next(gen))  #Output: 1

print(next(gen))  #Output: 2

print(next(gen))  #Output: 3
# print(next(gen))  #Raises Stop Iteration


Q7.What are the advantages of using generators over regular functions?

Answer.

Advantages of Using Generators over Regular Functions:

Generators offer several powerful benefits, especially when dealing with large datasets or when performance and memory efficiency are important. Here’s a detailed breakdown:

#1. Memory Efficiency

Generators do not store the entire sequence in memory.

They yield items one by one, which is ideal for large or infinite sequences.

#2. Lazy Evaluation

Values are computed only when needed, not in advance.

This improves performance in situations where you don't need all results at once.

#3. Improved Performance for Iteration

Since values are generated on-the-fly, there's less overhead than storing entire results.

Ideal for streaming data or processing files line by line.

**Example.**

def counter(max):

    current = 0
    while current < max:
        yield current
        current += 1

Q9. Explain the purpose and usage of the `map()` function in Python.


Answer.

The map() function is a built-in Python function used to apply a given function to each item of an iterable (like a list or tuple) and return a new iterable (a map object) with the results.

**Syntax:**

map(function, iterable)

function: A function (can be a regular function or lambda) that is applied to each item.

iterable: A sequence (like a list, tuple, or set) whose elements you want to transform.

**Example :**

**Using map() with a Lambda Function**

numbers = [1, 2, 3, 4, 5]

squares = map(lambda x: x ** 2, numbers)

print(list(squares))  # Output: [1, 4, 9, 16, 25]



Q10.. What is the difference between `map()`, `reduce()`, and `filter()` functions in Python?

Answer.

These are all higher-order functions used for processing iterables, but each serves a different purpose.

#1 map() - Transform Elements

Applies a function to each item in an iterable and returns a new iterable with the transformed items.

**Syntax:**

map(function, iterable)

**Example**

nums = [1, 2, 3, 4]

result = map(lambda x: x ** 2, nums)

print(list(result))  # Output: [1, 4, 9, 16]


#2. filter() - Select Elements


Filters elements of an iterable by applying a function that returns True or False.

**Syntax:**

filter(function, iterable)

**Example:**

nums = [1, 2, 3, 4, 5]

result = filter(lambda x: x % 2 == 0, nums)

print(list(result))  # Output: [2, 4]

#3. reduce() - Aggregate Elements

Applies a function to reduce the iterable to a single value (e.g., sum, product).

**Syntax:**

from functools import reduce

reduce(function, iterable)

 **Example:**

from functools import reduce

nums = [1, 2, 3, 4]

result = reduce(lambda x, y: x + y, nums)

print(result)  # Output: 10

```python
# 1. Write a Python function that takes a list of numbers as input and returns the sum of all even numbers in
#the list.
def sum_even_numbers(numbers):
    even_sum = 0
    for num in numbers:
        if num % 2 == 0:
            even_sum += num
    return even_sum

my_list = [1, 2, 3, 4, 5, 6]
result = sum_even_numbers(my_list)
print("Sum of even numbers:", result)


```

```python
#4. Write a Python function that checks if a given number is prime or not from 1 to 200.
def is_prime(n):
    if n <= 1:
        return False  # 0 and 1 are not prime
    for i in range(2, int(n**0.5) + 1):  # check up to square root of n
        if n % i == 0:
            return False
    return True
for num in range(1, 201):
    if is_prime(num):
        print(num, "is a prime number")

```

```python
#5.. Create an iterator class in Python that generates the Fibonacci sequence up to a specified number of
#terms.
class FibonacciIterator:
    def __init__(self, max_terms):
        self.max_terms = max_terms
        self.count = 0
        self.a = 0
        self.b = 1

    def __iter__(self):
        return self

    def __next__(self):
        if self.count >= self.max_terms:
            raise StopIteration
        if self.count == 0:
            self.count += 1
            return self.a
        elif self.count == 1:
            self.count += 1
            return self.b
        else:
            self.a, self.b = self.b, self.a + self.b
            self.count += 1
            return self.b
fib = FibonacciIterator(10)  # Generate 10 terms
for num in fib:
    print(num, end=" ")
```

```python
#6.Write a generator function in Python that yields the powers of 2 up to a given exponent.
def powers_of_two(max_exponent):
    for i in range(max_exponent + 1):
        yield 2 ** i
for power in powers_of_two(5):
    print(power)
```

```python
#7. Implement a generator function that reads a file line by line and yields each line as a string.
def read_file_line_by_line(file_path):
    with open(file_path, 'r') as file:
        for line in file:
            yield line.strip()  # strip() removes leading/trailing whitespace including newline
for line in read_file_line_by_line('example.txt'):
    print(line)
```

```python
#10. Create a Python program that uses `filter()` to remove all the vowels from a given string.
def remove_vowels(input_str):
    vowels = 'aeiouAEIOU'
    return ''.join(filter(lambda char: char not in vowels, input_str))

# Example usage
text = "Hello, World!"
result = remove_vowels(text)
print("String without vowels:", result)
```
